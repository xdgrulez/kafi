# Appendix

## kwargs

These are the currently supported `kwargs` for the operators of `TopologyNode` (and therefore, also for `Streams`):
* `pack_fun: r -> hashable_r`: transforms/"packs" a the input record (typically a dictionary) to a hashable record for the underlying pydbsp engine. Default: `msgpack.packb`.
* `unpack_fun: hashable_r -> r`: transforms back/"unpacks" a hashable record obtained from the underlying pydbsp engine to an output record (typically a dictionary). Default: `msgpack.unpackb`.
* `to_zSet_fun: list(any) -> zSet`: transforms a list of input records of any type (typically, a list of dictionaries) to a pydbsp ZSet. Default: `TopologyNode: 
* `from_zSet_fun: zSet -> r`: transforms back an output ZSet obtained from the underlying pydbsp engine to a list of output records of any type.

## Naming conventions and typing

For simplicity (and because I am a fan of my own kind of "Hungarian notation" anyway from good old [Mozart](http://mozart2.org/) days ;-)), Kafi Streams doesn't use "real" Python type annotations but a homebrewed variable type naming/suffixing system.

This is a (probably incomplete) list of these variable type names/suffixes, in alphabetical order:
| Name/suffix | Description |
| :--- | :--- |
| `any` | any type |
| `bool` | Python `function` |
| `built_tn` | built `TopologyNode` object (and, since `Streams` is a subclass of `TopologyNode`, also a built `Streams` object) |
| `dict` | Python `dict` |
| `fun` | Python `function` |
| `int` | Python `int` |
| `kwargs` | Python `kwargs` |
| `list` | Python `list` |
| `m` | "message" (also a Python dictionary). This is a special kind of record = a Kafka message dictionary with `key`, `value`, `headers` etc. In broad strokes, `r` (see below) is used inside a Kafi Streams topology, and `m` outside |
| `r` | "record" or "row" in Kafi Streams (=typically a Python dictionary, but can also be just a tuple or a string etc.) |
| `set` | Python `set` |
| `str` | Python `str` |
| `tn` | object of the Kafi Streams `TopologyNode` class (and, since `Streams` is a subclass of `TopologyNode`, also a `Streams` object) |
| `tuple` | Python `tuple` |
| `w` | pydbsp weight (=Python `int`) |
