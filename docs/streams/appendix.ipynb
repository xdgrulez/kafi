## Naming conventions and typing

For simplicity (and because I am a fan of my own kind of "Hungarian notation" anyway from good old [Mozart](http://mozart2.org/) days ;-)), Kafi Streams doesn't use "real" Python type annotations but a homebrewed variable type naming/suffixing system.

This is a list of these variable type names/suffixes, in alphabetical order:
| Name/suffix | Description |
| :--- | :--- |
| `any` | any type |
| `bool` | Python `function` |
| `built_tn` | built `TopologyNode` object (and, since `Streams` is a subclass of `TopologyNode`, also a built `Streams` object) |
| `dict` | Python `dict` |
| `evaluator` | pydbsp `Evaluator` object |
| `fun` | Python `function` |
| `int` | Python `int` |
| `iterable` | Python `Iterable` |
| `kwargs` | Python `kwargs` |
| `list` | Python `list` |
| `m` | "message" (also a Python dictionary). This is a special kind of record = a Kafka message dictionary with `key`, `value`, `headers` etc. In broad strokes, `r` (see below) is used inside a Kafi Streams topology, and `m` outside |
| `nodeId` | pydbsp node ID (=Python `int`) |
| `r` | "record" or "row" in Kafi Streams (=typically a Python dictionary, but can also be just a tuple or a string etc.) |
| `set` | Python `set` |
| `str` | Python `str` |
| `tn` | object of the Kafi Streams `TopologyNode` class (and, since `Streams` is a subclass of `TopologyNode`, also a `Streams` object) |
| `ts` | timestamp (= Python `int`)
| `tuple` | Python `tuple` |
| `w` | pydbsp weight (=Python `int`) |
| `zSet` | pydbsp `ZSet` object |

## Keyword arguments (kwargs)

This section is about the keyword arguments (`kwargs`) of the *TopologyNode* and *Streams* classes.


<a id="operators"></a>
### Operators

This is the list of currently supported `kwargs` for the Kafi Streams operators, all related to [pydbsp integration](pydbsp.ipynb):
* [pack_fun()](#pack_fun) Set the `pack_fun` of this operator. Default: `msgpack.unpackb`
* [unpack_fun()](#unpack_fun) Set the `unpack_fun` of this operator. Default: `lambda x: msgpack.unpackb(x, strict_map_key=False)`
* [to_zSet_fun()](#to_zset_fun) Set the `to_zSet_fun` of this operator. Default: `from_records`
* [from_zSet_fun()](#from_zset_fun) Set the `from_zSet_fun` of this operator. Default: `to_records`
