# Streams

Welcome to the documentation the *Streams* part of *Kafi Streams*.

To get started, just go to the [Quickstart](quickstart.ipynb).

Learn about the architecture behind Kafi Streams in [Architecture](architecture.ipynb)

The operators of Kafi Streams are explained in [Operators](operators.ipynb) - and a separate chapter is dedicated to Time Operators.

How data enters Kafi Streams is laid out in Ingress, how data is processed and exits it in Processing/egress.

Lifecycle elucidates how to build a Kafi Streams topology and run it.

Checkpointing shows how Kafi Streams ensures fault tolerance.

And to avoid bugs in the first place, Debugging is about how to test drive and visualize Kafi Streams topologies.
