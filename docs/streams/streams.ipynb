# Streams

Welcome to the documentation the *Streams* part of *Kafi Streams*.

To get started, just go to the [Quickstart](quickstart.ipynb).

Learn about the architecture behind Kafi Streams in [Architecture](architecture.ipynb)

All the operators of Kafi Streams are explained in [Operators](operators.ipynb).

How data enters, is processed and exits Kafi Streams is laid out in [Processing](processing.ipynb).

[Lifecycle](lifecycle.ipynb) shows the options for building a Kafi Streams topology and running it.

[Checkpoints](checkpoints.ipynb) describes how Kafi Streams ensures fault tolerance.

And to avoid bugs in the first place, [Debugging](debugging.ipynb) is about how to test drive and visualize Kafi Streams topologies.

The [Appendix](appendix.ipynb) includes further information about additional keyword arguments, and naming conventions and typing.
