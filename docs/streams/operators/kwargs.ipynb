# Keyword arguments


These are the currently supported `kwargs` for the operators of `TopologyNode` (and therefore, also for `Streams`):
* `pack_fun: r -> hashable_r`: transforms/"packs" a the input record (typically a dictionary) to a hashable record for the underlying pydbsp engine. Default: `msgpack.packb`.
* `unpack_fun: hashable_r -> r`: transforms back/"unpacks" a hashable record obtained from the underlying pydbsp engine to an output record (typically a dictionary). Default: `msgpack.unpackb`.
* `to_zSet_fun: list(any) -> zSet`: transforms a list of input records of any type (typically, a list of dictionaries) to a pydbsp ZSet. Default: `TopologyNode: 
* `from_zSet_fun: zSet -> r`: transforms back an output ZSet obtained from the underlying pydbsp engine to a list of output records of any type.
