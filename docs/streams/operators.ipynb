# Operators

This chapter is about the operators of Kafi Streams.

It begins with the simpler [stateless operators](operators/stateless.ipynb) and continues with the more complex [stateful ones](operators/stateful.ipynb).

Then we turn to the "boss operators": [time operators](operators/time.ipynb), and close with the more low-level [weighted operators](operators/weighted.ipynb).

The chapter is very much example-driven because we are convinced that examples say more than 100 words ;-)

We will base our examples on the `TopologyNode` class directly instead of `Streams` to avoid the overhead of the latter and Kafka. Hence, you can try them all out without Kafka.

You'll notice that all operators have additional keyword arguments, listed in the [appendix](appendix.ipynb).
