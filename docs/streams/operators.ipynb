# Operators

This chapter is about the operators of Kafi Streams.

It begins with the [stateless operators](operators/stateless.ipynb) and continues with the [stateful ones](operators/stateful.ipynb), 

Then we turn to the "boss operators": [time operators](operators/time.ipynb), and close with the more low-level [weighted operators](operators/weighted.ipynb).

This chapter is very much example-driven because we are convinced that examples say more than 100 words ;-) We will base our examples on the `TopologyNode` class directly instead of `Streams` to avoid the overhead of the latter (and Kafka). For brevity, we will also use an implicit sink definition (only possible with `TopologyNode`, not with `Streams` where explicit sinks are obligatory).

You'll notice that all operators have additional [keyword arguments](operators/kwargs.ipynb) (`kwargs`).
