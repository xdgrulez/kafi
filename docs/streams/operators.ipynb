# Operators

This chapter is about the operators of Kafi Streams. It doesn't cover the operators concerning time and time windows (for that, see Time Operators).

It begins with stateless operators, goes over to the stateful ones, and then goes deeper into DBSP/pydbsp with weighted operators and DBSP operators. You probably won't ever need the latter two.

Since we intend to show you the operation of each of the operators by example, we'll use the `TopologyNode` class instead of `Streams` (see [Architecture](architecture.ipynb) for understanding the relationship between `TopologyNode` and `Streams`).

We also use an implicit sink definition (only possible with `TopologyNode`, not with `Streams`) for brevity.
